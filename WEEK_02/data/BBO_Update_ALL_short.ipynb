```python
import os
import numpy as np

# === 0. Base directory where all function folders live ===
# If your notebook is in that folder, this is fine:
base_dir = os.getcwd()

print("Base directory:", base_dir)
print("Contents:", os.listdir(base_dir))
```

```python
def show_file_info(folder, tail=1):
    """Load initial_* for a folder and print shape + last `tail` values."""
    inputs_path  = os.path.join(folder, "initial_inputs.npy")
    outputs_path = os.path.join(folder, "initial_outputs.npy")

    inputs_data  = np.load(inputs_path)
    outputs_data = np.load(outputs_path)

    print(f"\n{folder} SHAPE - Inputs:", inputs_data.shape,"Outputs:", outputs_data.shape)

    # last `tail` rows / values (handles if there are fewer than `tail`)
    print(f"Last {min(tail, len(inputs_data))} inputs:", inputs_data[-tail:], f"Last {min(tail, len(outputs_data))} outputs:", outputs_data[-tail:] )
    # print(f"Last {min(tail, len(outputs_data))} outputs:\n", outputs_data[-tail:])

# call for F1..F8
for i in range(1, 9):
    show_file_info(f"F{i}", tail=1)
```

```python
def update_and_show(folder, new_x_row, new_y_value, save_as_initial=False):
    """
    Load initial_inputs/outputs for `folder`,
    append ONE new row/value,
    save updated arrays,
    and print shapes + last values.
    """

    inputs_path  = os.path.join(folder, "initial_inputs.npy")
    outputs_path = os.path.join(folder, "initial_outputs.npy")

    # --- load old data ---
    inputs_data  = np.load(inputs_path)
    outputs_data = np.load(outputs_path)

    # --- prepare new data ---
    new_x_row   = np.array(new_x_row).reshape(1, -1)     # (1, D)
    new_y_value = np.array([new_y_value])                # (1,)

    # safety: dimension must match
    assert new_x_row.shape[1] == inputs_data.shape[1], (
        f"{folder}: new_x_row dim {new_x_row.shape[1]} != existing dim {inputs_data.shape[1]}"
    )

    # --- append ---
    inputs_updated  = np.append(inputs_data, new_x_row, axis=0)
    outputs_updated = np.append(outputs_data, new_y_value)

    # --- save ---
    if save_as_initial:
        # overwrite original files
        np.save(inputs_path,  inputs_updated)
        np.save(outputs_path, outputs_updated)
    else:
        # save as new files
        np.save(os.path.join(folder, "updated_inputs.npy"),  inputs_updated)
        np.save(os.path.join(folder, "updated_outputs.npy"), outputs_updated)

    # --- print info ---
    print(f"\n{folder} SHAPE before/after - Inputs: {inputs_data.shape} -> {inputs_updated.shape}, "
          f"Outputs: {outputs_data.shape} -> {outputs_updated.shape}")
    print("Last values - input:",  inputs_updated[-1:], "output:", outputs_updated[-1:])

    return inputs_updated, outputs_updated
```
