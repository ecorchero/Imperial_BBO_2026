```python
import numpy as np

# Load the file
data = np.load("initial_inputs.npy")

# Show last row before change
print("Last row BEFORE change:")
print(data[-1])

# Modify the last row
data[-1] = [0.010200, 0.988600, 0.990900]

# Save back to the SAME file (overwrites it)
np.save("initial_inputs.npy", data)

# Reload to verify it was saved
data_check = np.load("initial_inputs.npy")

print("\nLast row AFTER change (reloaded from disk):")
print(data_check[-1])

```
